```python
from astropy.coordinates import SkyCoord
import astropy.units as u
from astroquery.simbad import Simbad
from astroquery.ipac.ned import Ned
import warnings
warnings.filterwarnings('ignore') # Suppresses standard astropy warnings for clean output

# 1. Define the coordinates from the Legacy Survey tooltip
ra_val = 187.1729
dec_val = 15.0007

# Create an astropy SkyCoord object
coord = SkyCoord(ra=ra_val*u.deg, dec=dec_val*u.deg, frame='icrs')

# Define a tight search radius (5 arcseconds is standard for cross-matching faint galaxies)
search_radius = 5 * u.arcsec

print(f"Searching for objects within {search_radius} of RA={ra_val}, Dec={dec_val}...\n")

# 2. Query SIMBAD
print("--- SIMBAD Database ---")
try:
    # Customizing SIMBAD to fetch morphological types and redshifts if available
    custom_simbad = Simbad()
    custom_simbad.add_votable_fields('otype', 'z_value')
    
    simbad_result = custom_simbad.query_region(coord, radius=search_radius)
    
    if simbad_result is not None:
        # Displaying the most relevant columns
        print(simbad_result[['MAIN_ID', 'RA', 'DEC', 'OTYPE', 'Z_VALUE']])
    else:
        print("No objects found in SIMBAD at these coordinates.")
except Exception as e:
    print(f"SIMBAD query failed: {e}")

print("\n")

# 3. Query NED (NASA/IPAC Extragalactic Database)
print("--- NED (Extragalactic Database) ---")
try:
    ned_result = Ned.query_region(coord, radius=search_radius)
    
    if ned_result is not None:
        # NED has specific column names, we'll pull the ones most relevant for galaxies
        columns_to_show = ['Object Name', 'RA', 'DEC', 'Type', 'Velocity', 'Redshift']
        print(ned_result[columns_to_show])
    else:
        print("No objects found in NED at these coordinates.")
except Exception as e:
    print(f"NED query failed: {e}")
```
