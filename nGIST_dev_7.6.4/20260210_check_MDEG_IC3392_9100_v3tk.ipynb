```python
from pathlib import Path
import os
import warnings
import numpy as np
from astropy.io import fits
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from concurrent.futures import ThreadPoolExecutor, as_completed

warnings.filterwarnings("ignore", message="All-NaN slice encountered", category=RuntimeWarning)

# -----------------------------
# USER SETTINGS
# -----------------------------
base = Path.cwd()

run_ids = {
    8:  "IC3392_v3tk_9100_mask_uptoCONT_MDEG8",
    12: "IC3392_v3tk_9100_mask_uptoCONT_MDEG12",
    14: "IC3392_v3tk_9100_mask_uptoCONT_MDEG14",
    16: "IC3392_v3tk_9100_mask_uptoCONT_MDEG16",
    20: "IC3392_v3tk_9100_mask_uptoCONT_MDEG20",
}

linecube_name = lambda rid: base / rid / f"{rid}_line_cube.fits"

ORIG_CUBE_PATH = Path("/Users/Igniz/nGISTENV/ngist_supplementary_public/ngistTutorial/inputData/IC3392_DATACUBE_FINAL_WCS_Pall_mad_red_v3tk.fits")

windows = {
    "cont_5050_5120": (5050.0, 5120.0),
    "cont_5218_5485": (5218.0, 5485.0),
    "cont_5600_5650": (5600.0, 5650.0),
    "cont_6000_6200": (6000.0, 6200.0),
    "cont_6400_6500": (6400.0, 6500.0),
    "Hb_edge_4800_4920": (4800.0, 4920.0),
    "OIII_edge_4940_5060": (4940.0, 5060.0),
    "HaNII_edge_6500_6625": (6500.0, 6625.0),
    "SII_edge_6680_6760": (6680.0, 6760.0),
}

diff_pair = (20, 14)      # difference map: A - B
NSAMPLE  = 20000         # spaxels used for stats/hists
STRIDE   = 4             # spatial downsample for diff map
NCPU     = min(os.cpu_count() or 8, 16)  # cap threads to avoid I/O thrash
REF_MDEG = 14                     # baseline for percentage comparisons
DENOM_FLOOR_FRAC = 1e-3           # protects % plots when denominator ~ 0
CONT_FLOOR_FRAC = 1e-3            # protects CONT-normalized plots when continuum ~ 0
CONT_WIN = (5218.0, 5485.0)       # continuum window for normalization



# -----------------------------
# FITS HELPERS
# -----------------------------
def wave_from_fits_header(hdr):
    crval = hdr["CRVAL3"]
    cdelt = hdr.get("CDELT3", hdr.get("CD3_3"))
    crpix = hdr.get("CRPIX3", 1.0)
    n = hdr["NAXIS3"]
    pix = np.arange(n) + 1
    return crval + (pix - crpix) * cdelt

def open_memmap_cube(path):
    hdul = fits.open(path, memmap=True)
    cube = hdul[0].data  # (nwave, ny, nx)
    wave = wave_from_fits_header(hdul[0].header)
    return hdul, cube, wave

def open_memmap_cube_anyhdu(path: Path):
    hdul = fits.open(path, memmap=True)
    hdu = None
    for hh in hdul:
        if getattr(hh, "data", None) is None:
            continue
        if hh.data is None:
            continue
        if hh.data.ndim == 3:
            hdu = hh
            break
    if hdu is None:
        raise RuntimeError(f"No 3D cube found in {path}")
    cube = hdu.data
    wave = wave_from_fits_header(hdu.header)
    return hdul, cube, wave

def robust_rms_1d(x):
    x = x[np.isfinite(x)]
    if x.size == 0:
        return np.nan
    med = np.median(x)
    mad = np.median(np.abs(x - med))
    return 1.4826 * mad

def continuum_level_map(cube, wave, w1, w2):
    idx = (wave >= w1) & (wave <= w2)
    if not np.any(idx):
        raise RuntimeError("Continuum window has no samples on this wavelength grid.")
    return np.nanmedian(cube[idx, :, :], axis=0)



# -----------------------------
# LOAD CUBES ONCE
# -----------------------------
hduls, cubes, waves = {}, {}, {}
ref_mdeg = sorted(run_ids.keys())[0]

for mdeg, rid in run_ids.items():
    f = linecube_name(rid)
    hdul, cube, wave = open_memmap_cube(f)
    hduls[mdeg] = hdul
    cubes[mdeg] = cube
    waves[mdeg] = wave
    print(f"[LOAD] MDEG={mdeg:>2}  file={f.name}  shape={cube.shape}")

ref_shape = cubes[ref_mdeg].shape
ref_wave  = waves[ref_mdeg]
for mdeg in sorted(run_ids):
    if cubes[mdeg].shape != ref_shape:
        raise ValueError(f"Shape mismatch: MDEG {mdeg} {cubes[mdeg].shape} vs {ref_shape}")
    if not np.allclose(waves[mdeg], ref_wave, atol=1e-6, rtol=0):
        raise ValueError(f"Wavelength grid mismatch at MDEG {mdeg}")

# Load the ORIGINAL cube for continuum normalization
hdul_orig, cube_orig, wave_orig = open_memmap_cube_anyhdu(ORIG_CUBE_PATH)
print(f"[LOAD] ORIG cube: {ORIG_CUBE_PATH.name}  shape={cube_orig.shape}")

# Only require spatial agreement (ny, nx). Spectral axis can differ.
if cube_orig.shape[1:] != ref_shape[1:]:
    raise ValueError(f"ORIG cube spatial shape mismatch: {cube_orig.shape[1:]} vs {ref_shape[1:]}")

# Build continuum level map from the ORIGINAL cube using its own wavelength grid
cont_map = continuum_level_map(cube_orig, wave_orig, *CONT_WIN)
cont_finite = np.isfinite(cont_map)
cont_scale = np.nanpercentile(np.abs(cont_map[cont_finite]), 90)
CONT_FLOOR = cont_scale * CONT_FLOOR_FRAC
print(f"[CONT] p90(|cont|)={cont_scale:.4g}, floor={CONT_FLOOR:.4g}")



# -----------------------------
# BUILD A RANDOM SPAXEL SAMPLE (using a reference window)
# -----------------------------
w1_ref, w2_ref = windows["cont_5218_5485"]
idx_ref = (ref_wave >= w1_ref) & (ref_wave <= w2_ref)
sub_ref = cubes[ref_mdeg][idx_ref, :, :]
valid_ref = np.isfinite(sub_ref).any(axis=0)

yy, xx = np.where(valid_ref)
n_valid = yy.size
if n_valid == 0:
    raise RuntimeError("No valid spaxels found in reference window; check mask/cube.")

rng = np.random.default_rng(42)
take = min(NSAMPLE, n_valid)
sel_idx = rng.choice(n_valid, size=take, replace=False)
ys = yy[sel_idx]
xs = xx[sel_idx]
print(f"[SAMPLE] Using {take} spaxels out of {n_valid} valid spaxels (ref {w1_ref}-{w2_ref} Å)")
print(f"[PARALLEL] Using {NCPU} threads")

# Continuum normalization for sampled spaxels
cont_samp = cont_map[ys, xs]
cont_samp_denom = np.sign(cont_samp) * np.maximum(np.abs(cont_samp), CONT_FLOOR)



# -----------------------------
# FAST SAMPLE-BASED STATS (fixed indexing)
# -----------------------------
def compute_window_stats_for_run(mdeg, w1, w2):
    cube = cubes[mdeg]
    wave = ref_wave

    idx = (wave >= w1) & (wave <= w2)
    if not np.any(idx):
        return mdeg, None, np.full(take, np.nan), np.full(take, np.nan)

    slab = cube[idx, :, :]              # (nwave_win, ny, nx)
    sub  = slab[:, ys, xs]              # (nwave_win, nsample)

    valid = np.isfinite(sub).any(axis=0)
    med_all = np.full(take, np.nan, dtype=float)
    rms_all = np.full(take, np.nan, dtype=float)

    if valid.sum() == 0:
        return mdeg, {"n_valid": 0, "n_tot": int(valid.size)}, med_all, rms_all

    subv = sub[:, valid]

    # per-spaxel median
    med = np.nanmedian(subv, axis=0)

    # per-spaxel robust rms (MAD along wavelength)
    mad = np.nanmedian(np.abs(subv - med[None, :]), axis=0)
    rms = 1.4826 * mad

    med_all[valid] = med
    rms_all[valid] = rms

    stats = {
        "n_valid": int(valid.sum()),
        "n_tot": int(valid.size),
        "med_p16": float(np.nanpercentile(med, 16)),
        "med_p50": float(np.nanpercentile(med, 50)),
        "med_p84": float(np.nanpercentile(med, 84)),
        "rms_p16": float(np.nanpercentile(rms, 16)),
        "rms_p50": float(np.nanpercentile(rms, 50)),
        "rms_p84": float(np.nanpercentile(rms, 84)),
    }
    return mdeg, stats, med_all, rms_all



def compute_diff_map(w1, w2, mdeg_a, mdeg_b, stride=4):
    idx = (ref_wave >= w1) & (ref_wave <= w2)
    if not np.any(idx):
        return None

    sub_a = cubes[mdeg_a][idx, ::stride, ::stride]
    sub_b = cubes[mdeg_b][idx, ::stride, ::stride]
    return np.nanmedian(sub_a - sub_b, axis=0)



# -----------------------------
# MAIN WINDOW LOOP
# -----------------------------
for wname, (w1, w2) in windows.items():
    print("\n" + "="*90)
    print(f"WINDOW: {wname}  ({w1:.1f}–{w2:.1f} Å)")

    med_all_by_run = {}
    rms_all_by_run = {}
    stats_by_run = {}

    with ThreadPoolExecutor(max_workers=NCPU) as ex:
        futures = [ex.submit(compute_window_stats_for_run, mdeg, w1, w2) for mdeg in sorted(run_ids)]
        for fut in as_completed(futures):
            mdeg, stats, med_all, rms_all = fut.result()
            stats_by_run[mdeg] = stats
            med_all_by_run[mdeg] = med_all
            rms_all_by_run[mdeg] = rms_all

    # ----- absolute summary -----
    for mdeg in sorted(run_ids):
        s = stats_by_run.get(mdeg, None)
        if s is None:
            print(f"  MDEG={mdeg:>2}: window has no wavelength samples")
            continue
        if s.get("n_valid", 0) == 0:
            print(f"  MDEG={mdeg:>2}: no valid sampled spaxels ({s.get('n_valid',0)}/{s.get('n_tot','?')})")
            continue
        print(
            f"  MDEG={mdeg:>2}: "
            f"med(p16,p50,p84)=({s['med_p16']:.4g},{s['med_p50']:.4g},{s['med_p84']:.4g})  "
            f"rms(p16,p50,p84)=({s['rms_p16']:.4g},{s['rms_p50']:.4g},{s['rms_p84']:.4g})  "
            f"sample_valid={s['n_valid']}/{s['n_tot']}"
        )

    # ----- RMS as % of continuum -----
    print(f"\n  RMS/CONT [%] (continuum window {CONT_WIN[0]:.0f}-{CONT_WIN[1]:.0f} Å):")
    for mdeg in sorted(run_ids):
        rms = rms_all_by_run[mdeg]
        ok = np.isfinite(rms) & np.isfinite(cont_samp_denom)
        if ok.sum() == 0:
            print(f"    MDEG={mdeg:>2}: no valid spaxels for RMS/CONT")
            continue
        rms_pct = rms[ok] / np.abs(cont_samp_denom[ok]) * 100.0
        p16, p50, p84 = np.nanpercentile(rms_pct, [16, 50, 84])
        print(f"    MDEG={mdeg:>2}: RMS/CONT [%] (p16,p50,p84)=({p16:.3g}, {p50:.3g}, {p84:.3g})")

    # ----- % increase in RMS relative to REF_MDEG -----
    ref_rms = rms_all_by_run[REF_MDEG]
    ref_ok = np.isfinite(ref_rms) & (ref_rms > 0)

    print(f"\n  RMS % increase relative to MDEG {REF_MDEG}:")
    pct_by_run = {}
    for mdeg in sorted(run_ids):
        if mdeg == REF_MDEG:
            continue
        rms = rms_all_by_run[mdeg]
        ok = ref_ok & np.isfinite(rms)
        if ok.sum() == 0:
            print(f"    MDEG={mdeg:>2}: no overlap-valid spaxels for % comparison")
            pct_by_run[mdeg] = np.array([])
            continue
        pct = (rms[ok] / ref_rms[ok] - 1.0) * 100.0
        pct_by_run[mdeg] = pct
        p16, p50, p84 = np.nanpercentile(pct, [16, 50, 84])
        print(f"    MDEG={mdeg:>2}: ΔRMS% (p16,p50,p84)=({p16:.3g}, {p50:.3g}, {p84:.3g})   N={ok.sum()}")

    # ----- % histogram overlays (RMS increase) -----
    plt.figure(figsize=(8, 4))
    for mdeg in sorted(run_ids):
        if mdeg == REF_MDEG:
            continue
        pct = pct_by_run.get(mdeg, np.array([]))
        if pct.size == 0:
            continue
        plt.hist(pct, bins=80, histtype="step", density=True, label=f"MDEG {mdeg} vs {REF_MDEG}")
    plt.axvline(0, ls="--", lw=1)
    plt.title(f"Residual RMS change (%): {wname}")
    plt.xlabel(f"ΔRMS [%] relative to MDEG {REF_MDEG}")
    plt.ylabel("PDF (normalized)")
    plt.legend()
    plt.show()

    # ----- optional: median-residual histogram -----
    plt.figure(figsize=(8, 4))
    for mdeg in sorted(run_ids):
        vals = med_all_by_run.get(mdeg, np.array([]))
        if vals is None or np.all(~np.isfinite(vals)):
            continue
        vv = vals[np.isfinite(vals)]
        plt.hist(vv, bins=80, histtype="step", density=True, label=f"MDEG {mdeg}")
    plt.title(f"Median residual distribution (sampled spaxels): {wname}")
    plt.xlabel("Per-spaxel median residual in window (LINE cube)")
    plt.ylabel("PDF (normalized)")
    plt.legend()
    plt.show()

    # ----- map product: % change in RMS map (stable) -----
    def window_rms_map(mdeg, w1, w2, stride=4):
        idx = (ref_wave >= w1) & (ref_wave <= w2)
        if not np.any(idx):
            return None
        slab = cubes[mdeg][idx, ::stride, ::stride]
        med = np.nanmedian(slab, axis=0)
        mad = np.nanmedian(np.abs(slab - med[None, :, :]), axis=0)
        return 1.4826 * mad

    a, b = diff_pair
    rms_a = window_rms_map(a, w1, w2, stride=STRIDE)
    rms_b = window_rms_map(b, w1, w2, stride=STRIDE)

    if rms_a is not None and rms_b is not None and np.isfinite(rms_b).any():
        finite_b = np.isfinite(rms_b) & (rms_b > 0)
        if finite_b.any():
            floor = np.nanpercentile(rms_b[finite_b], 5) * DENOM_FLOOR_FRAC
            denom = np.maximum(rms_b, floor)
            pct_map = (rms_a - rms_b) / denom * 100.0

            finite = np.isfinite(pct_map)
            v = np.nanpercentile(np.abs(pct_map[finite]), 99) if finite.any() else 10.0

            plt.figure(figsize=(6, 5))
            plt.imshow(pct_map, origin="lower", vmin=-v, vmax=v)
            plt.colorbar(label=f"ΔRMS[%] = 100*(RMS{a}-RMS{b})/RMS{b}  (stride={STRIDE})")
            plt.title(f"% RMS-change map: {wname}")
            plt.xlabel("x [pix/stride]")
            plt.ylabel("y [pix/stride]")
            plt.show()



# -----------------------------
# SPAXEL SPECTRUM OVERLAY (x,y from Mapviewer)
# -----------------------------
XPIX = 218  # set from Mapviewer
YPIX = 218  # set from Mapviewer

def plot_spaxel_suite(
    x, y,
    mdegs=None,
    diff_pair=(20, 14),
    wmin=4800, wmax=7000,
    sharey=True,
    ):
    """
    One figure with:
      - N panels: individual MDEG spectra
      - 1 panel: overlay of all MDEGs
      - 1 panel: difference spectrum (A-B)
    Bottom two panels span both columns (true spanning using GridSpec).
    """
    if mdegs is None:
        mdegs = tuple(sorted(run_ids))

    w = ref_wave
    sel = (w >= wmin) & (w <= wmax)

    # ---- layout: dynamic rows for individual panels, last two rows span both columns ----
    n_ind = len(mdegs)
    ncols = 2
    nrows_ind = int(np.ceil(n_ind / ncols))
    fig = plt.figure(figsize=(14, 2.6 * (nrows_ind + 2)), constrained_layout=True)
    gs = GridSpec(
        nrows=nrows_ind + 2,
        ncols=ncols,
        figure=fig,
        height_ratios=([1] * nrows_ind) + [1.1, 1.1],
    )

    axes_ind = []
    for i, mdeg in enumerate(mdegs):
        r = i // ncols
        c = i % ncols
        ax = fig.add_subplot(gs[r, c], sharey=axes_ind[0] if sharey and axes_ind else None)
        axes_ind.append(ax)

    # overlay spans full row
    ax_overlay = fig.add_subplot(gs[nrows_ind, :], sharex=axes_ind[0] if axes_ind else None)

    # diff spans full row
    ax_diff = fig.add_subplot(gs[nrows_ind + 1, :], sharex=axes_ind[0] if axes_ind else None)

    # ---- plot individual runs ----
    for ax, mdeg in zip(axes_ind, mdegs):
        spec = cubes[mdeg][:, y, x]
        good = sel & np.isfinite(spec)
        ax.plot(w[good], spec[good], lw=1)
        ax.axhline(0, ls="--", lw=1)
        ax.set_title(f"LINE cube: MDEG {mdeg}  (x={x}, y={y})")
        ax.set_ylabel("Flux")

    # x-label only on bottom-most axes to reduce clutter
    for ax in axes_ind:
        ax.label_outer()

    # ---- overlay panel ----
    for mdeg in mdegs:
        spec = cubes[mdeg][:, y, x]
        good = sel & np.isfinite(spec)
        ax_overlay.plot(w[good], spec[good], lw=1, label=f"MDEG {mdeg}")
    ax_overlay.axhline(0, ls="--", lw=1)
    ax_overlay.set_title(f"Overlay: all MDEGs  (x={x}, y={y})")
    ax_overlay.set_ylabel("Flux")
    ax_overlay.legend(ncol=min(len(mdegs), 4), loc="upper left")

    # ---- difference panel (% of local continuum) ----
    cont0 = cont_map[y, x]
    denom0 = np.sign(cont0) * max(abs(cont0), CONT_FLOOR) if np.isfinite(cont0) else CONT_FLOOR

    ref_mdeg = diff_pair[1]
    db = cubes[ref_mdeg][:, y, x]

    for mdeg in mdegs:
        if mdeg == ref_mdeg:
            continue
        da = cubes[mdeg][:, y, x]
        d_abs = da - db
        d_pct_cont = d_abs / denom0 * 100.0
        good = sel & np.isfinite(d_pct_cont)
        if good.any():
            ax_diff.plot(w[good], d_pct_cont[good], lw=1, label=f"MDEG {mdeg}-{ref_mdeg}")

    ax_diff.axhline(0, ls="--", lw=1)
    ax_diff.set_title(
        f"Difference (% of continuum): 100*(LINE(MDEG)-LINE(MDEG{ref_mdeg}))/CONT  (x={x}, y={y})"
        f"\nCONT median({CONT_WIN[0]:.0f}-{CONT_WIN[1]:.0f}) = {cont0:.4g}"
    )
    ax_diff.set_xlabel("Wavelength [Å]")
    ax_diff.set_ylabel("Δ Flux / CONT [%]")
    ax_diff.legend(ncol=min(3, len(mdegs)), loc="lower right")

    plt.show()

def plot_spaxel_diff(x, y, mdeg_a=20, mdeg_b=14, wmin=4800, wmax=7000):
    w = ref_wave
    sel = (w >= wmin) & (w <= wmax)
    a = cubes[mdeg_a][:, y, x]
    b = cubes[mdeg_b][:, y, x]
    d = a - b
    good = sel & np.isfinite(d)
    plt.figure(figsize=(10, 3))
    if good.sum() > 0:
        plt.plot(w[good], d[good], lw=1)
    plt.axhline(0, ls="--", lw=1)
    plt.xlabel("Wavelength [Å]")
    plt.ylabel(f"LINE(MDEG{mdeg_a}) - LINE(MDEG{mdeg_b})")
    plt.title(f"Spaxel difference spectrum at (x={x}, y={y})")
    plt.show()

plot_spaxel_suite(
    XPIX, YPIX,
    diff_pair=(20, 14),
    wmin=4800, wmax=7000,
    sharey=True,
    )

# Optional: zoom to "continuum-ish" window
plot_spaxel_suite(XPIX, YPIX, wmin=5218, wmax=5485)

# Optional: zoom to Hα+[N II]
plot_spaxel_suite(XPIX, YPIX, wmin=6500, wmax=6625)

# plot_spaxel_diff(XPIX, YPIX, diff_pair[0], diff_pair[1], 4800, 7000)


```

```python
# -----------------------------
# LINE-FLUX INTEGRATED CHANGES (per spaxel sample)
# -----------------------------
def dw_from_header(hdr):
    return float(hdr.get("CDELT3", hdr.get("CD3_3")))

DW = abs(dw_from_header(hduls[REF_MDEG][0].header))

line_windows = {
    "Hb":   (4847.0, 4876.0),
    "OIII": (5000.0, 5025.0),
    "HaNII":(6545.0, 6585.0),
    "SII":  (6710.0, 6745.0),
}

for lname, (w1, w2) in line_windows.items():
    print("\n" + "-"*70)
    print(f"{lname} ({w1}-{w2} Å): integrated flux changes")

    idx = (ref_wave >= w1) & (ref_wave <= w2)
    if idx.sum() == 0:
        print("  (no wavelength samples in this window)")
        continue

    # Vectorized integration for ALL sampled spaxels at once
    # slab shape: (nwave_in_window, nsample)
    slab_ref = cubes[REF_MDEG][idx][:, ys, xs]
    Fref = np.nansum(slab_ref, axis=0) * DW
    ok_ref = np.isfinite(Fref) & (np.abs(Fref) > 0)

    for mdeg in sorted(run_ids):
        if mdeg == REF_MDEG:
            continue

        slab = cubes[mdeg][idx][:, ys, xs]
        F = np.nansum(slab, axis=0) * DW

        ok = ok_ref & np.isfinite(F)
        if ok.sum() == 0:
            print(f"  MDEG {mdeg}: no overlap-valid spaxels")
            continue

        dF_pct = (F[ok] - Fref[ok]) / Fref[ok] * 100.0
        p16, p50, p84 = np.nanpercentile(dF_pct, [16, 50, 84])
        print(f"  MDEG {mdeg}: ΔF/Fref [%] (p16,p50,p84)=({p16:.3g}, {p50:.3g}, {p84:.3g})  N={ok.sum()}")

        # ΔF/CONT scaling (CONT from original cube)
        cont_ok = ok & np.isfinite(cont_samp_denom)
        if cont_ok.any():
            dF_cont_pct = (F[cont_ok] - Fref[cont_ok]) / np.abs(cont_samp_denom[cont_ok]) * 100.0
            c16, c50, c84 = np.nanpercentile(dF_cont_pct, [16, 50, 84])
            print(f"            ΔF/CONT [%] (p16,p50,p84)=({c16:.3g}, {c50:.3g}, {c84:.3g})  N={cont_ok.sum()}")

# -----------------------------
# REQUIRED ZOOM-INS AROUND KEY LINES
# -----------------------------
plot_spaxel_suite(XPIX, YPIX, wmin=4800, wmax=4920)  # Hβ
plot_spaxel_suite(XPIX, YPIX, wmin=4940, wmax=5060)  # [O III]
plot_spaxel_suite(XPIX, YPIX, wmin=6500, wmax=6625)  # Hα+[N II]
plot_spaxel_suite(XPIX, YPIX, wmin=6680, wmax=6760)  # [S II]

# Close only when you're fully done:
for hdul in hduls.values():
    hdul.close()
hdul_orig.close()
```

```python
# -----------------------------
# OVERFIT DIAGNOSTIC: inferred continuum model from ORIG cube
# -----------------------------
def interp_to_refwave(spec_in, wave_in, wave_out):
    ok = np.isfinite(spec_in) & np.isfinite(wave_in)
    if ok.sum() < 2:
        return np.full_like(wave_out, np.nan, dtype=float)
    return np.interp(wave_out, wave_in[ok], spec_in[ok], left=np.nan, right=np.nan)

def local_continuum_from_sidebands(orig_on_ref, wave, sidebands):
    # sidebands: list of (w1, w2)
    m = np.zeros_like(wave, dtype=bool)
    for w1, w2 in sidebands:
        m |= (wave >= w1) & (wave <= w2)
    vals = orig_on_ref[m]
    vals = vals[np.isfinite(vals)]
    if vals.size == 0:
        return np.nan
    return np.nanmedian(vals)

# Example sidebands (adjust if needed)
SIDE = {
    "Hb":   [(4825, 4842), (4882, 4900)],
    "OIII": [(4985, 4995), (5032, 5050)],
    "Ha":   [(6505, 6535), (6595, 6620)],
    "SII":  [(6685, 6705), (6750, 6770)],
}

def plot_overfit_diagnostic(x, y, line_name="Ha", wmin=6500, wmax=6625):
    w = ref_wave
    sel = (w >= wmin) & (w <= wmax)

    # original spectrum on ref_wave
    orig_ref = interp_to_refwave(cube_orig[:, y, x], wave_orig, ref_wave)

    # normalization: local continuum near the line (better than a fixed blue window)
    cont_local = local_continuum_from_sidebands(orig_ref, ref_wave, SIDE[line_name])
    denom = np.nanmax([np.abs(cont_local), CONT_FLOOR])

    plt.figure(figsize=(12, 6))

    # Panel A: inferred continuum models (this is the key!)
    ax1 = plt.subplot(2, 1, 1)
    ax1.plot(w[sel], orig_ref[sel], lw=1, label="ORIG")
    for mdeg in sorted(run_ids):
        line = cubes[mdeg][:, y, x]
        cont_model = orig_ref - line
        ax1.plot(w[sel], cont_model[sel], lw=1, label=f"CONT_model (MDEG {mdeg})")
    ax1.set_ylabel("Flux")
    ax1.set_title(f"{line_name}: ORIG vs inferred continuum models at (x={x},y={y})")
    ax1.legend(ncol=2, fontsize=9)

    # Panel B: residuals as % of local continuum (absolute, not difference)
    ax2 = plt.subplot(2, 1, 2, sharex=ax1)
    for mdeg in sorted(run_ids):
        line = cubes[mdeg][:, y, x]
        ax2.plot(w[sel], 100.0 * line[sel] / denom, lw=1, label=f"LINE/CONT% (MDEG {mdeg})")
    ax2.axhline(0, ls="--", lw=1)
    ax2.set_ylabel("LINE / CONT [%]")
    ax2.set_xlabel("Wavelength [Å]")
    ax2.legend(ncol=2, fontsize=9)

    plt.tight_layout()
    plt.show()

# Example usage:
plot_overfit_diagnostic(XPIX, YPIX, line_name="Hb", wmin=4800, wmax=4920)
plot_overfit_diagnostic(XPIX, YPIX, line_name="OIII", wmin=4940, wmax=5060)
plot_overfit_diagnostic(XPIX, YPIX, line_name="Ha", wmin=6500, wmax=6625)
plot_overfit_diagnostic(XPIX, YPIX, line_name="SII", wmin=6680, wmax=6760)
```
